```python
# Define a layer (like Dense in TF)
# All this is to calculate the output (Activation) values 
# It will not be training the model

def layer(a_in, W, b, g):
    units = W.shape[1]        # 
    a_out = np.zeros(units)
    for j in range(units):             
        w = W[:,j]                     
        z = np.dot(w, a_in) + b[j]     
        a_out[j] = g(z)                
    return a_out
```

```python

def sigmoid(z):
    return 1/(1+np.exp(-z))

def softmax(z):
    return np.exp(z)/(np.sum(np.exp(z)))

def my_model(x, W1, b1, W2, b2, W3, b3):
    a1 = layer(x,W1,b1,sigmoid)
    a2 = layer(a1,W2,b2,sigmoid)
    a3 = layer(a2,W3,b3,softmax)
    return a3

    
```
